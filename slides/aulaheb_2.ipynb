## Crescimento

Vamos começar perguntando: o que sabemos do crescimento do Brasil no século XIX? As estimativas dos níveis e do crescimento do PIB neste período são precárias e controversas.

As estimativas de Celso Furtado (1959) indicam um longo
período de estagnação na 1a metade do século XIX, mas
significativa expansão na 2a metade daquele século.

Crescimento da renda *per capita* – 1800-50: próximo a zero;
1850-1900: Brasil = 1,5% a.a.; Nordeste = -0,6% a.a.; Sul = 1,0%
a.a.; Sudeste = 2,3% a.a. 

Segundo Furtado, a estagnação na 1a metade do século seria consequência do declínio das exportações (0,8% a.a.) frente ao crescimento da população (1,3% a.a.); na 2a metade, houve acentuada recuperação das exportações, café em particular.

## Estimativas de Goldsmith

Já as estimativas de Richard Goldsmith (1986) identificaram um crescimento muito menor da renda *per capita* na segunda metade do século XIX: 0,3% a.a.

Uma figura similar foi estimada por Claudio Contador e Claudio
Haddad (1975) para 1861-1900: 0,4% a.a.

Angus Maddison (2001), usando os dados de Goldsmith, obteve resultados (em PPP) ainda menores: -0,02 a.a. entre 1850-1890. As estimativas de Goldsmith (via Maddison) são as mais utilizadas em comparações internacionais que incluem o Brasil.

Há novas estimativas sendo elaboradas atualmente por Edmar Bacha, Guilherme Tombolo e Flávio Versiani, mas ainda são muito preliminares. Elas apontam para maior crescimento na 2a metade do século XIX e menor crescimento no século XX.

Um estudo mais recente (2025, inédito) de Thales Pereira, com novos índices de preços por regiões, reavalia o crescimento do PIB durante a 2a metade do século XIX em 1,19% a.a. - portanto, um pouco menor, mas convergente com a estimativa original de Celso Furtado vista antes.

## Estimativas e seus problemas

Problema: estimativas tão divergentes entre Furtado e Goldsmith (e outros) levam a conclusões opostas com relação ao crescimento na segunda metade do século XIX. Qual é mais plausível? São confiáveis?

As estimativas de Furtado são baseadas nas exportações e na
população de várias regiões (1959, caps. 19 e 25).

Já as estimativas de Goldsmith resultam de uma média dos i) índices de salários pagos no Rio de Janeiro (Lobo, 1978), ii) X + M, iii) gastos governamentais da União e iv) do agregado monetário M2 encadeados com as estimativas de produto de Claudio Haddad para 1910.

## Exportações e indústria

O estudo de Warren Dean, *A industrialização de São Paulo* (1969), foi pioneiro ao mostrar que a indústria no Brasil cresceu rapidamente com a enorme expansão do café no século XIX e início do século XX (e *não somente* na década de 1930). Dean notou na época:

> “costuma-se afirmar, por estranho que pareça, que o desenvolvimento da indústria não ocorreu do crescimento do comércio do café, mas do seu gradativo colapso”.

## Choque externo?

Nessa última interpretação (defendida até então por vários autores brasileiros, CEPAL e corroborada recentemente por Gómez-Galvarriato & Williamsom), a indústria teria surgido e se expandido como resultado de choques externos ou crises do setor exportador.

Porém, as evidências compiladas nas últimas décadas mostram uma realidade diferente, de estreita relação entre setor exportador/café e indústria.

São Paulo do século XIX e início do século XX seria um caso bem sucedido da *Staples thesis* (Innis, 1930, 1937; North, 1955, 1959; Watkins, 1963)?

## Máquinas e equipamentos

Outra questão importante diz respeito às primeiras indústrias que surgiram na esteira do *boom* exportador. 

A visão tradicional (teoria cepalina e outras por ela influenciada, que continuam influentes) é de que a indústria de máquinas e equipamentos *somente surgiu na década de 1930, ou mais tarde ainda, na década de 1950*.

No entanto, pesquisas antigas (Leff, 1968; Dean, 1969; Lago et al., 1979) e recentes (Marson, 2012) mostraram que oficinas e fundições iniciaram a produção de máquinas e equipamentos em São Paulo desde o final do século XIX.

## 2a Parte - Ferrovia

Vamos passar a agora para os desenvolvimentos da infraestrutura
e das políticas públicas durante o período de rápida diversificação
industrial no Brasil.

As primeiras leis para criação de ferrovias foram a) Lei Imperial de
1835 – ligação norte-sul do país; b) Lei Provincial de São Paulo, 1836 –
Santos-capital. Nenhuma delas deu origem a companhias.

Possíveis causas do insucesso inicial:

- falta de empreendedorismo? Ausência de "visão empresarial para localizar e combinar [os] recursos escassos" - K, L, tecnologia.

- Taxa de retorno esperada baixa e incerta, instituições?

## Espírito empreendedor x instituições
    
Na primeira legislação de 1835, p.ex., havia previsão de tarifas
máximas, sem garantias quanto aos retornos.

Para Summerhill (2003) a explicação para o fracasso das
primeiras iniciativas no setor ferroviário residiria no desenho
institucional do programa, que não criou incentivos claros e
seguros para atrair investimentos privados.

Já a hipótese de ausência de visão empresarial atribui,
implicitamente, o insucesso das primeiras iniciativas ao baixo
nível de desenvolvimento econômico, o que somente foi superado com
a expansão da economia cafeeira.

## Nova legislação

O esquema bem-sucedido: Lei no. 641 de 1852, com os
seguintes privilégios principais:

i) 5% de “garantia de juros” (dividendos das ações emitidas pelas companhias) assumida pelo governo imperial; a partir de 1855, com 2% extras pelos governos provinciais;

ii) poder de desapropriação de terras particulares (com indenização) e cessão gratuita de terras devolutas pelo governo imperial de áreas ocupadas pelas linhas ferroviárias, estações, armazéns e demais prédios;

iii) redução dos direitos de importação para material ferroviário (trilhos, máquinas e equipamentos) e carvão mineral;

iv) direito de extração de madeira em terras devolutas para as obras da ferrovia;

v) privilégio de zona, área de pouco mais de 30 km (5 léguas) em
ambos os lados da linha férrea; posteriormente, concessão de 90 anos.

## Primeiras ferrovias 

Primeiros projetos ferroviários. **Nacionais**: 

- 1852 (operação em 1854), Estrela-Petrópolis no Rio de Janeiro;

- 1858 (operação), Recife-São Francisco; 1875, E.F. D. Pedro II, Rio-Cachoeira (SP).

**São Paulo**, dois exemplos: 

- *Santos-Jundiaí*: concessão em 1856, incorporação da cia. em 1860 como *San Paulo Railway Company* (SPR). Concessão original vendida a investidores ingleses, início de operação em 1867. Garantia de juros: 5% governo imperial e 2% governo provincial, com reembolso aos governos imperial e provincial com retorno acima de 8%. Ligação do porto de Santos ao interior;

## Interesses e negócios

Ao contrários de outras províncias no Brasil, as ferrovias em São
Paulo foram todas incorporadas por capitalistas locais (exceto a
SPR, que foi transferida para investidores ingleses).

Famílias ricas e influentes participavam de várias atividades: café
(produção, comércio, imigração, financiamento), ferrovia e
governo.

Por exemplo, acionistas da Cia. Paulista (Silva Prado, Souza
Queiroz, Vergueiro e Paes de Barros) ocupavam funções no
governo (p.ex., a Sociedade Promotora de Imigração) e em bancos
(Souza Queiroz, p.ex.).

## Garantia de juros

Sistema de garantia de juros lançado pela lei geral de 1852 foi bem-sucedido, atraindo capital tanto doméstico quanto estrangeiro.

*Garantia de juros*: quando a rentabilidade da companhia ficava abaixo da taxa de retorno anual contratada (p. ex., 7%), governo repassava a diferença (que *era debitada no passivo* da empresa, dívida); 

quando o retorno era maior do que a taxa cotratada (p. ex., 8%), o excedente era *repassado ao governo para reembolso* das garantias adiantadas. Terminado o reembolso, o excedente era *dividido entre governo e companhia*.

Companhias paulistas obtiveram taxas de retorno médias anuais superiores às taxas pagas pelos títulos do governo (entre 5 e 7% a.a.). 

As principais companhias abandonaram voluntariamente o esquema de garantia de juros assim que concluíram o reembolso das subvenções recebidas, tendo em vista a rentabilidade elevada e consistente dos projetos.

## Impactos

Impactos da ferrovia:

- *tradicionalmente citados*: extensão da fronteira do café; ocupação de regiões distantes; mobilização de capital; renovou estruturas gerenciais, financeiras e tecnológicas; fator de dissolução da sociedade escravista por ser “elemento estranho” na sociedade escravista (será?);

## Encadeamentos

Apesar de relativamente reduzidos (em comparação a países europeus), os efeitos diretos e indiretos da instalação das ferrovias sobre a produção de insumos, equipamentos e treinamento da força de trabalho foram (possivelmente) importantes na expansão inicial da indústria.

As oficinas estabelecidas pela Cia Paulista e a Mogiana, por exemplo, possuíam fundição e manufatura de peças, carros, vagões e, no início do século XX, locomotivas.

Além de utilizarem equipamentos e insumos importados em grande escala, essas empresas empenharam-se em aprendizado, adaptação e produção própria de material ferroviário de vários tipos. Esses aspectos da indústria ferroviária, porém, ainda são pouco conhecidos.

## Social savings

O principal impacto positivo das ferrovias foi via *social savings*, isto é, reduções de custo, tempo e risco produzidas pelo transporte de mercadorias e pessoas, liberando recursos para outros usos alternativos e beneficiando a economia como um todo.

Os ganhos ou *social savings* dependem dos meios de transportes alternativos existentes e da importância das mercadorias transportadas no produto da economia.

No caso do Brasil do século XIX, o alto custo do transporte existente permitiu que as ferrovias gerassem uma grande economia de recursos, ampliando mercados, elevando a mobilidade do trabalho e criando novas oportunidades econômicas. Impacto direto, portanto, sobre níveis de produto e renda.

## Educação primária

Outro tema relevante é a relação entre instituições da educação primária e desenvolvimento econômico – aspecto que tem sido considerado importante para explicar as trajetórias divergentes de crescimento de longo prazo nas

## Constituição e educação

A primeira Constituição do Brasil ([Constituição de 1824](www.planalto.gov.br/ccivil_03/Constituicao/Constituicao24.htm), art. xxxii) estabeleceu o princípio geral da educação elementar gratuita como um direito de todos os cidadãos (o que excluía os
escravos, por definição) e sob responsabilidade das *províncias*. Porém, nenhum mecanismo de financiamento foi
estabelecido.

O [Ato Adicional de 1834](https://www2.camara.leg.br/legin/fed/leimp/1824-1899/lei-16-12-agosto-1834-532609-publicacaooriginal-14881-pl.html) reafirmou o princípio constitucional da responsabilidade da educação primária
para as províncias, mas novamente sem provisão financeira específica para
governos provinciais ou municipais.

Desta forma, os níveis de gasto público com educação primária e matrículas dependiam das decisões das assembléias provinciais – e, portanto, das elites que as controlavam.

## Elites e centralização

Este arranjo institucional no Brasil foi distinto da descentralização observada em outros países líderes na instrução primária no século XIX. 


Nos Estados Unidos e Alemanha, por exemplo, as escolas primárias eram financiadas e administradas localmente, com parcela expressiva de impostos sobre propriedade (sistema já adotado em colônias britânicas sob o regime de *self-government*).

No Brasil, câmaras municipais não dispunham de recursos nem autonomia fiscal ou de criação de escolas. Portanto, alto nível de centralização administrativa e financeira:

- no governo central (que na prática concentrou-se no ensino superior, exceto na capital do Império, cidade do Rio de Janeiro); 

- nos governos e assembleias provinciais (ensino primário e secundário). 

Aparentemente, a elevada centralização foi também diferente de outros países da América Latina em boa parte do século XIX (Engerman, Mariscal and Sokoloff). Vejamos as taxas de matrículas de alguns países - notar que os resultados são bastante distintos em termos absolutos e considerando o PIB *per capita*.

## Desigualdade regional

Dentro do Brasil, a situação nas províncias também era bastante desigual em termos educacionais.

Em 1872, a taxa de matrícula do Brasil era de 14 crianças por
1.000 habitantes. Os líderes educacionais eram: Rio de Janeiro, capital do Império (31), Sergipe (29), Rio Grande do Sul (23), Santa Catarina (21), Espírito Santo (21), Amazonas (21).

Entre as menores taxas estavam Piauí (8), Minas Gerais (9) e

(18), Maranhão (16) e Alagoas (15), por exemplo, apresentavam

## Visão das elites

Uma possível explicação para os ínfimos níveis educacionais no
Brasil seria a falta de consciência do problema. Mas o tema foi
amplamente debatido no século XIX e início do século XX.

Em São Paulo, pelo menos a partir dos anos 1850, as autoridades provinciais rotineiramente argumentaram que o “desenvolvimento intelectual” da província estava muito atrasado em relação ao seu “progresso econômico”.

Ainda assim, a visão predominante era a de que “em vista de outras despesas essenciais, é impossível aumentar os recursos destinados à educação” (Presidente da Província de SP, 1875).

## Tributos e receitas

Casos de tributação *local e específica* para a instrução primária foram raros no Brasil. Por exemplo, a primeira legislação com financiamento específico para a educação primária em São Paulo foi aprovada em 1887, inspirada nos exemplos de EUA e Alemanha.

Nessa lei de 1887, maiores de 21 anos pagariam um imposto escolar (*poll tax*; portanto, *não* sobre propriedade da terra ou outros ativos). Os recursos seriam administrados por um conselho nas municipalidades (nomeado pelas câmaras locais e governo). 

Porém, mesmo limitados e relativamente bem-sucedidos, os conselhos e o imposto escolar foram suprimidos pelo novo governo republicano em 1892.

Daí em diante, os recursos destinados à educação em SP seriam oriundos do orçamento fiscal ordinário. Com o aumento das receitas do café, a educação recebeu investimentos crescentes, tanto municipais quanto estaduais.

## Exportações e educação

A primeira Constituição (1891) da República atendeu à demanda por descentralização política e fiscal das elites cafeeiras, destinando o imposto de exportação aos estados.

Para o estado de São Paulo, isso significou um *aumento real de 7,5 vezes* nas receitas fiscais de 1905 em relação aos anos 1888-1889.

Com esses novos recursos, São Paulo tornou-se já em 1920 um dos líderes na oferta de educação pública no
Brasil, conforme figura.

```python
import matplotlib.pyplot as plt
import pandas as pd

matsp = pd.read_excel('fig_matriculasp.xlsx')

# Criar figura
fig_matricula, ax = plt.subplots(figsize=(8, 5))

# Preparar dados
x = range(len(matsp))
width = 0.35

# Criar barras
bars1 = ax.bar([i - width/2 for i in x], matsp['Brasil'], width, label='Brasil', color='#1f77b4')
bars2 = ax.bar([i + width/2 for i in x], matsp['São Paulo'], width, label='São Paulo', color='#ff7f0e')

# Configurações
ax.set_ylabel('Crianças matriculadas por 1.000 hab', fontsize=10)
ax.set_title('Taxa de matrícula no ensino primário (% de crianças por 1.000 hab), SP e BR', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(matsp['Ano'])

# Ajustar legenda - bbox_to_anchor controla a posição
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.08), ncol=2, 
          title='País/estado', frameon=True, fontsize=10)

plt.tight_layout()
plt.close(fig_matricula)  # Para não exibir automaticamente
```

## Exportações como explicação

Não só São Paulo, mas outros estados exportadores beneficiaram-se das maiores receitas fiscais, que permitiram maiores gastos em educação.

Assim, o aumento das exportações pode ter sido um elemento-chave para explicar a expansão na oferta de educação primária em alguns estados e o declínio em outros.

Mas há outros possíveis determinantes?

## Outros determinantes?

Outros elementos também podem ter sido importantes para explicar o pior ou melhor desempenho educacional, tais como
concentração da terra, imigrantes e urbanização. Estudos quantitativos recentes têm abordado essas questões.

No caso dos imigrantes, o efeito pode ter sido significativo sobre a oferta de educação primária em províncias/estados
que receberam um grande número de famílias europeias durante a grande imigração.

Os imigrantes europeus possuíam níveis mais elevados de educação do que a população brasileira local: na Alemanha, a taxa de matrícula entre 5-14 anos era de 70% na década de 1870; na Itália, 35% em 1890. No Brasil, 10% em 1900.

## Imigrantes e educação?

Há vários exemplos de mobilização dos imigrantes por escolas
primárias em São Paulo:

- primeiro, por meio de escolas comunitárias ou étnicas, formadas por imigrantes alemães e italianos, por exemplo. Em 1909, havia 115 escolas italianas em São Paulo (59 na Argentina, 80 nos EUA);

- segundo, por meio da pressão sobre autoridades locais para abertura de escolas públicas nas fazendas ou localidades. 

Estudos econométricos com municípios de São Paulo e Rio Grande do Sul têm corroborado o efeito positivo da presença de imigrantes na oferta de educação primária.

## Persistência

De qualquer forma, apesar do aumento das taxas de matrícula no Brasil no início do século XX, o atraso educacional frente a outros países persistiu.

Igualmente, o salto no desempenho educacional de São Paulo em relação a outros estados não foi suficiente para superar seu atraso educacional.

Não só as taxas de matrícula continuaram baixas em termos internacionais, como os níveis de analfabetismo permaneceram altíssimos, conforme mostram as 2 figuras a seguir.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

matpa = pd.read_excel('fig_matriculapaises.xlsx')

# Criar figura
fig_paises, ax = plt.subplots(figsize=(12, 6))

# Preparar dados
anos = matpa['Ano']
n_anos = len(anos)
paises = ['São Paulo', 'Brasil', 'Argentina', 'Uruguai', 'Portugal', 'Itália', 'Estados Unidos']
n_paises = len(paises)

# Posições e largura das barras
x = np.arange(n_anos)
width = 0.12  # Ajustado para 7 países

# Cores do colormap tab20
colors = plt.cm.tab20b(np.linspace(0, 0.7, n_paises))

# Criar barras para cada país
for i, pais in enumerate(paises):
    offset = (i - n_paises/2 + 0.5) * width
    ax.bar(x + offset, matpa[pais], width, label=pais, color=colors[i])

# Configurações
ax.set_ylabel('Crianças matriculadas por 1.000 hab', fontsize=12)
ax.set_title('Taxa de matrícula no ensino primário, 1870-1920 (% de crianças por 1.000 hab.)', fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(anos, fontsize=10)
ax.tick_params(axis='y', labelsize=10)

# Legenda mais próxima do gráfico
ax.legend(loc='upper center', 
          bbox_to_anchor=(0.5, -0.07),  # Ajuste este valor para controlar distância
          ncol=7,  # 4 colunas para melhor distribuição
          title='Países/estado',
          frameon=True,
          fontsize=11,
          title_fontsize=12,
          columnspacing=1.2)

ax.grid(axis='y', alpha=0.3, linestyle='--')

# Remover bordas
ax.spines['top'].set_visible(False)    # Remove linha superior
ax.spines['right'].set_visible(False)  # Remove linha direita
#ax.spines['left'].set_visible(False)   # Remove linha esquerda
#ax.spines['bottom'].set_visible(False) # Remove linha inferior

plt.tight_layout()
plt.close(fig_paises)
```
