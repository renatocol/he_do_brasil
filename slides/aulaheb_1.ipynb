Porém, em outras regiões o trabalho indígena

## Estrutura social

A estrutura social da economia do açúcar era
relativamente complexa, como na Bahia:


- *senhores de engenho e escravos*, mas também


- *lavradores* – brasileiros ou europeus (comerciantes, profissionais e militares) que plantavam cana para vender aos engenhos, sob contratos de 9-18 anos, pequenos proprietários de cativos (2-30 escravizados);

- *pobres* – brancos e libertos (isto é, ex-escravizados) que desempenhavam funções variadas, com salários anuais (mestre de engenho, supervisores), diários ou por tarefas (cortadores de lenha, carpinteiros).

## Outras atividades

A economia colonial do Nordeste não se resumia ao açúcar,
naturalmente. Outras atividades:

- *tabaco*: 2a mais importante até meados do século XVIII, reunindo pequenos mas também grandes produtores;

- *pecuária*: principalmente bovinos, utilizados no engenho, transporte e consumo; criação incluía grandes propriedades no sertão da Bahia, Pernambuco, Sergipe e outros, mas predominavam pequenas/médias propriedades;

- *mandioca*: i) produção de subsistência com pequenos excedentes levados a mercados locais; ii) produção em larga escala para engenhos e cidades.

## Economia do ouro

A partir do final do séc. XVII a busca e exploração de ouro
tornaram-se um empreendimento em larga escala, tanto em
Minas Gerais quanto em outras regiões (Mato Grosso e Goiás).

Ao tornar-se capitania independente em 1720, Minas Gerais
havia se consolidado como o principal centro da nova
atividade colonial.

A extração de ouro trouxe uma série de atividades produtivas
auxiliares. Exemplos:

- pecuária e agricultura de alimentos, tendo conexões com nordeste/sul do país;

- diversificação da atividade artesanal: carpinteiros, pedreiros, ferreiros;

- indústrias auxiliares: telhas, sabão, panelas e têxteis.

Maior período de expansão na extração do ouro ocorreu entre 1700 e 1720 (quase 5x); entre 1721 e 1735 continuou crescendo mas em menor ritmo; novo *boom* entre 1735 e 1750; a partir de 1750, lento declínio.

## Sentido da colonização

Como os historiadores interpretaram a Colônia? Uma visão clássica é a do “sentido da colonização” –
conceito utilizado por Caio Prado Jr. em sua análise da Colônia em seus livros *Formação do Brasil Contemporâneo – Colônia* (1942) e *História Econômica do Brasil* (1945).


> “Vasta empresa comercial [...] destinada a explorar os recursos naturais de um território ainda virgem em proveito do comércio europeu”.

>“É este o verdadeiro sentido da colonização tropical, de que o Brasil é uma das resultantes”.

## Visão clássica

Caio Prado estabeleceu o que é hoje considerada a visão
clássica da formação social e econômica brasileira como uma estrutura
produtiva com três elementos básicos:

- grande propriedade rural;

- monocultura;

- trabalho escravo.

Para Caio Prado, essa seria a “célula fundamental da estrutura agrária brasileira” e, mais do que isso, “a base principal em que se assenta toda a estrutura do país, econômica e social”.

Atividades voltadas ao mercado interno não possuiriam “vida autônoma”, tendo papel auxiliar às atividades fundamentais.

## Críticas e revisões

Vários trabalhos têm qualificado ou rejeitado a interpretação
da economia colonial de Caio Prado – que é às vezes chamada
de “visão plantacionista”. Dois aspectos são ilustrativos das
revisões propostas:

- *mercado interno*: a economia colonial poderia ser resumida à grande lavoura monocultora e exportadora? O setor agrícola doméstico seria de fato marginal e secundário?

- *propriedade escravista*: a unidade representativa da agricultura colonial seria de fato a grande propriedade escravista?

## Mercado interno e relações de trabalho

Estudos recentes têm mostrado que o mercado interno foi
ativo no Brasil-Colônia e no início do século XIX, com o
comércio de gêneros básicos (feijão, milho, charque, cereais,
sal) (Schwartz, Fragoso, p.ex.) (mas isso constitui de fato uma
novidade?).

Produção em pequenas propriedades e em terras arrendadas,
parceria e outras modalidades de contrato foram importantes.

Igualmente relevante foi a presença do trabalhador livre:
pequenos proprietários, agregados, arrendatários, camaradas
(trabalho por empreitada ou diarista, p.ex.) (Veremos esses tópicos na segunda parte desta aula).

## Propriedade escrava
    
Na economia do ouro em Minas Gerais o pequeno proprietário
de escravizados (até 5 cativos) foi dominante, com 70% ou mais
dos proprietários em várias localidades no início do século
XVIII (mas notar: maioria da população *não* possuía
escravizados).

Proprietários de até 5 escravizados detinham de 28% a 39% (de
acordo com o ano e localidade) do total e os com mais de 20
cativos variaram de 12,8% a 36,3% em 1804.

Padrão similar em outras regiões e épocas, embora com
variações. Por exemplo, no nordeste açucareiro (Schwartz) e
em MG e SP do início do século XIX (Luna & Klein).

Problema: como se define pequena ou grande propriedade?

## Controvérsias

## Setor exportador x doméstico?

Para Caio Prado Jr. e Celso Furtado, a agricultura de exportação
foi o setor dinâmico por excelência no sentido que induzia outras
atividades a ele relacionadas.

Ambos autores retrataram as pequenas e médias propriedades (e a
produção de alimentos) como “auxiliares”, “subsidiárias” ou
“dependentes” do setor agrícola exportador (dominado por
grandes propriedades).

Para eles, contudo, isso não significou em si domínio quantitativo:
p.ex., Furtado estimou que o setor exportador representava 1/4
da renda interna do Brasil em 1800, e 1/6 em 1850, o que implica
um vasto setor doméstico em termos de magnitude.

## 2a parte - Fim da escravidão

Vejamos agora o mercado de trabalho no Brasil após a

os anos 1810 (ainda sob Portugal), com destaque para a lei de [7 de

da escravidão passou a ser discutido em diversas ocasiões no
Parlamento, imprensa e encontros dos fazendeiros.

Brasil foi o maior importador de escravizados africanos no tráfico transatlântico, com mais de 5 milhões de africanos desembarcando em seus portos - contra menos de 300 mil nos Estados Unidos. Ver [estimativas](https://www.slavevoyages.org/assessment/estimates) e os portos de desembarque.

## Causas da abolição

Na historiografia brasileira, há duas hipóteses principais para o fim da escravidão em 1888. A *primeira* argumenta que a escravidão era economicamente ineficiente e que as elites cafeeiras no Rio de Janeiro e São Paulo possuíam mentalidades distintas:

a) no *Vale do Paraíba*, sul do Rio de Janeiro, a elite cafeeira seria
apegada à escravidão como forma de prestígio social em vez de
busca de lucro;

b) no *Oeste de São Paulo*, elite moderna, capitalista, dinâmica, aberta a novas idéias, em busca de novas relações de trabalho (imigração) e insatisfeita com a escravidão (O. Ianni, F.H. Cardoso, E. Viotti da Costa, Florestan Fernandes, Maria Isaura Queiroz).

## Ineficiência e elites

Assim, o surgimento de uma elite com mentalidade modernizante e capitalista no Oeste paulista teria levado à busca de novas relações de trabalho.

Com a urbanização e industrialização, a superioridade do trabalho
livre se tornaria evidente e entraria em conflito com a escravidão.

Portanto, nesta abordagem o declínio da demanda de escravizados (e queda dos preços) no período 1880-1888 se deveria sobretudo à
consciência da desvantagem econômica do trabalho escravo por parte das novas elites agrárias do Oeste paulista.

## Rentabilidade da escravidão

*Outra hipótese* considera que, no longo prazo, a escravidão estava condenada, como ficava claro com a Lei do Ventre Livre de 1871. No entanto, sustenta Pedro Carvalho de Mello (1977), “a abolição [1888] ocorreu bem antes do longo prazo”.

Segundo estimativas de Mello, no contrafactual de não ter havido a abolição em 1888 a população escrava
seria de 835.000 em 1892 e 544.000 em 1902.

Para Mello, a sobrevivência da escravidão nas regiões economicamente dinâmicas
seria possível via transferência escravizados de outras atividades para as fazendas de
café.

## Pressão abolicionista

Mello argumenta que a demanda de escravizados na década
de 1880 declinou devido a um fator exógeno: “pressão
abolicionista”.

Pressão abolicionista = crescente mobilização
antiescravista, campanhas, fuga das fazendas, redes de
apoio e fuga (p.ex., porto de Santos), que indicam rápida erosão
do apoio da sociedade à instituição escravista.

É esse o elemento principal que explicaria o declínio na
demanda de escravizados (e dos preços) no período,
segundo Mello.

## Eficiência x abolicionismo

A hipótese de Mello (1977), desta forma, é que o declínio da demanda ocasionado pela pressão abolicionista ocorreu *apesar* da viabilidade econômica da escravidão.

Uma das evidências em favor desta hipótese encontra-se na
comparação do mercado de compra e venda com o mercado de
aluguel de escravizados. A partir de 1881 (no RJ),

- o *preço de compra e venda dos escravizados declinou* acentuadamente, 


- enquanto o *preço do aluguel de escravizados se manteve estável*.

Ou seja, diante da crescente onda abolicionista, a incerteza quanto a
continuidade do regime servil derrubou a demanda pela compra
de escravizados e, por consequência, os seus preços. Mas os serviços dos
escravizados continuaram a ser requeridos, como demonstra a
estabilidade do preço do aluguel.

Esta seria uma evidência indireta, portanto, de que o emprego de escravizados *era ainda economicamente vantajoso* aos olhos dos proprietários rurais e urbanos.

## A ação direta dos escravizados

Dean (1977), por sua vez, dá outra ênfase: a *ação direta
dos escravizados*, desde o início da década de 1880, teria
sido a causa principal da derrocada da escravidão.

A ação direta dos escravizados teria se dado por meio de
desobediência, fugas, revoltas e uso da lei em seu favor.

Por outro lado, os fazendeiros apegaram-se até o último momento à escravidão, tentando postergar o seu fim. A
desorganização social provocada pelas ações dos escravizados levou ao aumento das alforrias, à adoção de
contratos de serviços a salário e, por fim, diante das fugas em massa, ao colapso do sistema.

## Povoamento e assalariamento

Vejamos agora outras relações de trabalho no Brasil do século XIX
*além da escravidão*. Os primeiros fluxos imigratórios (além dos
portugueses) para o Brasil ocorreram logo após a Independência
(1822). Dois modelos básicos prevaleceram desde então:

a) o *primeiro*, favorecido pelo governo imperial, que visava
estabelecer imigrantes europeus em colônias oficiais, em regime de pequenas
propriedades, sob justificativas que incluíam povoamento,
segurança de fronteiras, desenvolvimento da agricultura e
branqueamento da população. Casos mais notáveis no RS, SC, PR e
ES, mas também em outras províncias, como RJ e MG.

b) o *segundo modelo*, o de “colônias de parceria” e “colônias de
assalariamento”, predominou como política de imigração defendida
por fazendeiros e o governo provincial de São Paulo.

## Colônias de parceria

Vejamos o segundo modelo. As primeiras experiências com
imigrantes nas fazendas em São Paulo (região de Rio Claro/
Limeira no Oeste paulista) ocorreram na década de 1840, com alemães, suíços e
portugueses em regime de parceria – “colonos de parceria” (*sharecropping*).

Regime de parceria: “meio termo entre o regime de serviços
assalariados e o das pequenas propriedades” (S. Buarque de Holanda).

Nesse sistema, cada família cuidava de certo número de cafezais;
permitia-se cultivo de subsistência; produto da venda do café
colhido e de alimentos produzidos era repartido com o
fazendeiro. *Revolta de colonos suíços e alemães na fazenda Ibicaba em 1856*, crise do
sistema.

## Imigração subsidiada

A partir da Revolta de Ibicaba (1856) em Limeira, interior de São Paulo, o
sistema original de colonato de parceria entrou em declínio.
Contratos antigos foram redefinidos assim como os dos novos
imigrantes, embora muitas das características da parceria tenham
sido preservadas.

Estímulo da lei provincial no. 42 de 30/3/1871, com empréstimos a
fazendeiros para introdução de colonos europeus (norte da Europa,
preferencialmente). Novas leis em toda a década de 1880. A
inovação mais importante foi o pagamento das passagens dos
imigrantes europeus e Açores/Canárias – a “imigração subsidiada”,
lei no. 28, 29/3/1884.

O fluxo de entrada de imigrantes aumentou a partir de 1875, mas foi
especialmente de 1887 em diante que se observou o grande salto na
imigração européia em São Paulo. Ver tabela a seguir.

## Colônias de assalariamento

Novos contratos (sobretudo a partir da metade da década de 1880) deram
origem ao que se chamou de “colonato de assalariamento”.
Características dos novos contratos:

- contratos de 1 ano, colono e família trabalhavam como bem entendessem;

- remuneração fixa por tratamento de 1.000 pés de café, mais uma parte variável por quantidade de café colhido e outra por serviços ocasionais na fazenda;

- área de terra para cultivo de subsistência; em geral 2 hectares, com milho, arroz, feijão, batatas, legumes, com produto integral para os colonos; 

- colonos recebiam terreno para pasto, onde podiam criar galinhas, porcos, cabras e vacas, com o excedente podendo ser vendido.

O sucesso da política imigratória tendeu a ofuscar a presença de outras categorias de trabalhadores na lavoura paulista.

Mesmo assim, em outras regiões do Brasil (exceto os do sul do país)
os trabalhadores nacionais predominaram em vez de imigrantes
europeus, como veremos a seguir.

## Interpretações clássicas

Vejamos algumas interpretações sobre esses fatos. Uma forma
de abordar a questão do mercado de trabalho neste período foi
formulada por Celso Furtado:

- quais foram as razões para o uso maciço do trabalho imigrante em São Paulo – para Furtado, equivalente a Brasil – e da (presumida) não utilização do nacional? (Furtado, 1958, cap. xxi).

- Segundo Furtado, o setor de subsistência era representado nas fazendas por áreas ocupadas por agregados, que dispunham de uma roça e mantinham relação de dependência e subordinação frente ao grande proprietário rural.

## Leff e outros

Já a interpretação de Leff (1972) enfatizou outro aspecto:

- renda recebida pelos trabalhadores livres em suas atividades de subsistência (incluindo comércio de excedente) era superior ao custo dos escravizados;

- ou seja, “os salários necessários para atrair os trabalhadores livres do setor agrícola doméstico eram considerados relativamente altos” (dados?).

- Portanto, o emprego de escravizados seria economicamente vantajoso em relação aos trabalhadores livres com salários em um cenário contrafactual.

Outras interpretações clássicas: racismo, preconceito com
relação a “indolentes” e “vadios”, com aceitação passiva por
parte dos trabalhadores nacionais (Florestan Fernandes, Maria
S. Carvalho Franco).

## Interpretações recentes

Interpretações mais recentes têm enfatizado causas distintas
para a não utilização do trabalho nacional e o recurso ao
imigrante.

Uma delas enfatiza a atitude independente e autônoma do
trabalhador nacional, que rejeitaria as condições exigidas
(subordinação, privação de autonomia, disciplina) pelo
trabalho nas fazendas, preferindo formas precárias de
trabalho ou a marginalização (Hebe de Castro, Sheila Faria).

Outra interpretação sustenta que o trabalho nacional foi, na
verdade, empregado em escala apreciável (inclusive em São
Paulo), mas sujeito à sazonalidade do trabalho agrícola
(Lamounier).

## Outras regiões

Conforme mostrado por Stein (1957), no lugar de imigrantes europeus, os trabalhadores
nacionais livres e libertos tiveram participação expressiva no sul
do Rio de Janeiro como agregados, arrendatários, camaradas,
feitores e artesãos.

Em Pernambuco (Eisenberg, 1977), as experiências imigratórias
fracassaram apesar das tentativas oficiais. O que vingou foi a
mão de obra local, isto é, trabalhadores nacionais livres e
libertos.

Também em Minas Gerais o trabalhador nacional foi a mais
importante importante reserva de mão de obra. E assim foi com
a maioria dos outros estados – exceto os do Sul e Espírito Santo,
mas nesses, ainda assim, com imigrantes nas colônias oficiais.

## Excepcionalidade

Ou seja, São Paulo aparece mais como um *caso atípico* na
transição da escravidão para o trabalho livre no Brasil, por seu
emprego maciço de imigrantes europeus.

Como observou Lamounier (2007), porém, o caso de São
Paulo seria de fato um caso inteiramente atípico quanto
ao uso do trabalhador nacional?

Mesmo em São Paulo a presença de trabalhadores nacionais
pobres, livres e libertos, foi importante como fonte alternativa
ou complementar de trabalho.

## Diversidade

Categorias de trabalhadores nacionais (não imigrantes) em São
Paulo, assim como em outras províncias, antes e depois da
abolição:

- *pequenos proprietários e sitiantes*, que muitas vezes trabalhavam por empreitada ou temporariamente em fazendas vizinhas;


- *agregados*, com plantio de subsistência em lotes nas fazendas em troca de serviços;

- *camaradas*, contratados por empreitada ou como mensalistas, diaristas;

- *arrendatários (ou rendeiros)* de terras da fazenda;

- *parcerias*, em regime similar e simultâneo aos colonos imigrantes;

- locação de serviços (Lei de 1879), que previa contratos de 5 anos para o trabalhador nacional livre e 7 anos para o ex-escravizado.

Exemplo: em 1872, em Rio Claro, grande centro cafeeiro: número de
escravizados = 2.753; agregados e sitiantes = 1.700 (Dean).

Em resumo, mercado de trabalho diversificado e variado regionalmente: emprego maciço de imigrantes europeus limitado ao sul do Brasil e Espírito Santo (colônias oficiais) e São Paulo (colonos assalariados nas fazendas).

No restante do Brasil, predominou o trabalhador nacional livre e liberto.

## Quem eram os trabalhadores do café

Ainda assim, uma grande incógnita permanece sobre a importância *relativa* do trabalho escravo nas décadas anteriores à Abolição em maio de 1888, particularmente em São Paulo.

Com o boom do café, São Paulo tornou-se a principal região que demandava trabalho para as fazendas cafeeiras, importando escravizados de outras partes do país. 

Mas justamente pela intensidade da demanda de braços para o café, São Paulo também notabilizou-se pela demanda de imigrantes europeus, como vimos antes. E o mesmo, possivelmente, ocorreu com o trabalho nacional, incluindo libertos (ex-escravizados).

## Estimativas 

A escassez de dados e a falta de pesquisas mais sistemáticas dão margem para interpretações muito divergentes ou até opostas sobre a escravidão na década de 1880 - da perda de relevância do trabalho escravo ao domínio absoluto do trabalho imigrante (e irrelevância do trabalho nacional).

Uma estimativa recente, porém, chegamos a resultados mais precisos, embora ainda preliminares e tentativos.
